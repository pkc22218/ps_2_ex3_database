```python
# let's have a look at the columns in the game table
query = """
    PRAGMA table_info(draft_combine_stats)
"""
pd.read_sql(query, con)
```

```python
# let's find the earliest recorded game
query = """
    SELECT game_date
    FROM game
    ORDER BY game_date 
    LIMIT 1
"""

print(pd.read_sql(query, con))

# let's find the latest recorded game
query = """
    SELECT game_date
    FROM game
    ORDER BY game_date DESC
    LIMIT 1
"""

print(pd.read_sql(query, con))
```

```python
# TODO: Find four interesting facts about the data and write a query to find the answer
# Find the highest shoe height for the players
query = """
    SELECT 
        player_name,
        height_w_shoes,
        height_wo_shoes,
        height_w_shoes - height_wo_shoes
            AS shoe_height

    FROM draft_combine_stats
    ORDER BY shoe_height DESC
    LIMIT 10
"""

print(pd.read_sql(query, con))

```

```python
# Find the active players who might have the greatest potential if he loses weight
query = """
    SELECT 
        d.player_name,
        d.max_vertical_leap / d.body_fat_pct
            AS leap_per_body_fat

    FROM draft_combine_stats as d
    JOIN player as p
        ON p.id = d.player_id
    WHERE p.is_active = 1
    ORDER BY leap_per_body_fat DESC
    LIMIT 10
"""

print(pd.read_sql(query, con))
```

```python
# Find the teams with highest 5 teams with ave score after 2020
query = """
    SELECT 
        l.team_id_home,
        t.nickname,
        COUNT(*) as num_games,
        AVG(l.pts_home) as ave_score
    FROM line_score as l
    JOIN team_details as t
        ON l.team_id_home = t.team_id
    WHERE l.game_date_est >= '2020-01-01'
    GROUP BY l.team_id_home, t.nickname
    ORDER BY ave_score DESC
    LIMIT 5
"""

print(pd.read_sql(query, con))
```
